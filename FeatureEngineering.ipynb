```python
import pandas as pd

df = pd.read_csv(r"C:\Users\aniruddh.singh\OneDrive - Prodapt Solutions Private Limited\Documents\Project_prac\WA_Fn-UseC_-Telco-Customer-Churn(in).csv")
```

```python
df["churn_flag"] = df["Churn"].map({"Yes": 1, "No": 0})

df["tenure_bucket"] = pd.cut(
    df["tenure"],
    bins=[0, 12, 24, 48, 72],
    labels=["0-12", "13-24", "25-48", "49-72"],
    include_lowest=True
)
```

```python
df["tenure_bucket_code"] = df["tenure_bucket"].cat.codes

median_charge = df["MonthlyCharges"].median()

df["high_charge_flag"] = (
    df["MonthlyCharges"] > median_charge
).astype(int)
```

```python
service_cols = [
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]

for col in service_cols:
    df[col + "_flag"] = (df[col] == "Yes").astype(int)

df["service_count"] = df[
    [col + "_flag" for col in service_cols]
].sum(axis=1)
```

```python
df["is_long_term_customer"] = (
    df["tenure"] >= 24
).astype(int)
```

```python
df["has_streaming_bundle"] = (
    (df["StreamingTV"] == "Yes") &
    (df["StreamingMovies"] == "Yes")
).astype(int)
```

```python

df["auto_pay_flag"] = (
    df["PaymentMethod"]
    .str.contains("automatic", case=False, na=False)
    .astype(int)
)
```

```python
new_features = [
    "tenure_bucket_code",
    "high_charge_flag",
    "service_count",
    "is_long_term_customer",
    "has_streaming_bundle",
    "auto_pay_flag"
]

correlations = (
    df[new_features + ["churn_flag"]]
    .corr()["churn_flag"]
    .drop("churn_flag")
    .sort_values(key=abs, ascending=False)
)

print("Correlation of New Features with Churn")
print(correlations)
```

```python
df["tenure_bucket"] = pd.cut(
    df["tenure"],
    bins=[0, 12, 24, 48, 72],
    labels=["0-12", "13-24", "25-48", "49-72"],
    include_lowest=True
)
median_charge = df["MonthlyCharges"].median()

df["high_charge_flag"] = (
    df["MonthlyCharges"] > median_charge
).astype(int)

service_cols = [
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]

df["service_count"] = df[service_cols].eq("Yes").sum(axis=1)

df["is_long_term_customer"] = (
    df["tenure"] >= 24
).astype(int)

df["has_streaming_bundle"] = (
    (df["StreamingTV"] == "Yes") &
    (df["StreamingMovies"] == "Yes")
).astype(int)


df["auto_pay_flag"] = (
    df["PaymentMethod"]
      .str.contains("automatic", case=False, na=False)
      .astype(int)
)

customer_features = df.copy()

# Save to CSV
customer_features.to_csv(
    "customer_features.csv",
    index=False
)

print("Dataset saved as customer_features.csv")
print("Rows, Columns:", customer_features.shape)

print("\nNew columns added:")
print([
    "tenure_bucket",
    "high_charge_flag",
    "service_count",
    "is_long_term_customer",
    "has_streaming_bundle",
    "auto_pay_flag"
])
```
