```python
import pandas as pd

df = pd.read_csv(r"C:\Users\aniruddh.singh\OneDrive - Prodapt Solutions Private Limited\Documents\Project_prac\WA_Fn-UseC_-Telco-Customer-Churn(in).csv")
```

```python
contract_churn = (
    df.groupby("Contract")["Churn"]
      .apply(lambda x: (x == "Yes").mean() * 100)
      .sort_values(ascending=False)
      .reset_index(name="Churn_Rate_%")
)

print("\nChurn Rate by Contract Type")
print(contract_churn)

print(
    f"\nHighest churn contract: "
    f"{contract_churn.iloc[0]['Contract']} "
    f"({contract_churn.iloc[0]['Churn_Rate_%']:.2f}%)"
)
```

```python
internet_churn = (
    df.groupby("InternetService")["Churn"]
      .apply(lambda x: (x == "Yes").mean() * 100)
      .sort_values(ascending=False)
      .reset_index(name="Churn_Rate_%")
)

print("\nChurn Rate by Internet Service")
print(internet_churn)
```

```python
payment_churn = (
    df.groupby("PaymentMethod")["Churn"]
      .apply(lambda x: (x == "Yes").mean() * 100)
      .sort_values(ascending=False)
      .reset_index(name="Churn_Rate_%")
)

print("\nChurn Rate by Payment Method")
print(payment_churn)
```

```python
df["tenure_bucket"] = pd.cut(
    df["tenure"],
    bins=[0, 12, 24, 48, 72],
    labels=["0-12", "13-24", "25-48", "49-72"],
    include_lowest=True
)

tenure_churn = (
    df.groupby("tenure_bucket")["Churn"]
      .apply(lambda x: (x == "Yes").mean() * 100)
      .reset_index(name="Churn_Rate_%")
      .sort_values("Churn_Rate_%", ascending=False)
)

print("\nChurn Rate by Tenure Bucket")
print(tenure_churn)
```

```python
avg_monthly_charges = (
    df.groupby("Churn")["MonthlyCharges"]
      .mean()
      .reset_index(name="Avg_Monthly_Charges")
)

print("\nAverage Monthly Charges by Churn Status")
print(avg_monthly_charges)
```

```python
combo_churn = (
    df.groupby(["Contract", "InternetService"])["Churn"]
      .apply(lambda x: (x == "Yes").mean() * 100)
      .reset_index(name="Churn_Rate_%")
      .sort_values("Churn_Rate_%", ascending=False)
)

print("\nContract + Internet Service Churn Rates")
print(combo_churn)

highest_combo = combo_churn.iloc[0]

print("\nHighest Churn Combination")
print(
    f"Contract: {highest_combo['Contract']}, "
    f"Internet Service: {highest_combo['InternetService']}, "
    f"Churn Rate: {highest_combo['Churn_Rate_%']:.2f}%"
)
```
